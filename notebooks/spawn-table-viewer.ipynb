# Spawn Table Viewer

**Schema(s)** defining this data:
- `contracts/schemas/spawn-table.schema.json` — **SpawnTablePack**: schemaVersion, spawnIntervalTicks, capPerRoom, capPerLevel, entries[] with archetypeId, weight, minDepth, maxDepth.

**Asset / pack** we load:
- `contracts/data/content_spawn_table.json` — the spawn table pack

**What this tool does:** View spawn entries by depth band; rebalance weight and minDepth/maxDepth; behaviour = which archetype spawns where (assignment of archetypeId to depth/weight).

```python
import json
from pathlib import Path

ROOT = Path.cwd()
if not (ROOT / "packages" / "engine").is_dir() and (ROOT.parent / "packages" / "engine").is_dir():
    ROOT = ROOT.parent
CONTRACTS_DIR = ROOT / "packages" / "engine" / "src" / "escape-the-dungeon" / "contracts"
SCHEMA_PATH = CONTRACTS_DIR / "schemas" / "spawn-table.schema.json"
PACK_PATH = CONTRACTS_DIR / "data" / "content_spawn_table.json"

with open(PACK_PATH, encoding="utf-8") as f:
    SPAWN_PACK = json.load(f)

print("Schema:", SCHEMA_PATH.relative_to(ROOT))
print("Pack:  ", PACK_PATH.relative_to(ROOT))
print(f"Loaded spawnIntervalTicks={SPAWN_PACK.get('spawnIntervalTicks')} capPerRoom={SPAWN_PACK.get('capPerRoom')} capPerLevel={SPAWN_PACK.get('capPerLevel')}")
print(f"Entries: {len(SPAWN_PACK.get('entries', []))}")
```

## View spawn table

**Rebalance:** weight, minDepth, maxDepth in `content_spawn_table.json`. **Behaviour:** which archetypeId is eligible at which depth.

```python
def entries_at_depth(depth):
    return [e for e in SPAWN_PACK.get("entries", []) 
            if (e.get("minDepth") is None or e["minDepth"] <= depth) 
            and (e.get("maxDepth") is None or e["maxDepth"] >= depth)]

print("All entries (archetypeId, weight, minDepth, maxDepth):")
for e in SPAWN_PACK.get("entries", []):
    print(f"  {e.get('archetypeId', ''):12} weight={e.get('weight')} minDepth={e.get('minDepth')} maxDepth={e.get('maxDepth')}")

print("\nEligible at depth 1:")
for e in entries_at_depth(1):
    print(f"  {e['archetypeId']} weight={e['weight']}")
print("\nEligible at depth 12:")
for e in entries_at_depth(12):
    print(f"  {e['archetypeId']} weight={e['weight']}")
```
