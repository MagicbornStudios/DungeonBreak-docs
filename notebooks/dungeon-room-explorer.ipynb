# Dungeon and Room Explorer

**Schema(s)** defining this data:
- `contracts/schemas/dungeons.schema.json` — Dungeon structure (levels, rooms, room blueprints).
- `contracts/schemas/rooms.schema.json` — Room definitions.
- `contracts/schemas/room-templates.schema.json` — Room templates (room type / **room feature** = room type only).

**Assets / packs** we load:
- `contracts/data/content_dungeons.json` — dungeon pack
- `contracts/data/content_rooms.json` — rooms
- `contracts/data/content_room_templates.json` — room templates

**What this tool does:** View dungeons, floors, rooms; room type (room feature); exits. Rebalance room assignments; behaviour = which room has which room feature.

```python
import json
from pathlib import Path

ROOT = Path.cwd()
if not (ROOT / "packages" / "engine").is_dir() and (ROOT.parent / "packages" / "engine").is_dir():
    ROOT = ROOT.parent
CONTRACTS_DIR = ROOT / "packages" / "engine" / "src" / "escape-the-dungeon" / "contracts"
DATA = CONTRACTS_DIR / "data"

with open(DATA / "content_dungeons.json", encoding="utf-8") as f:
    DUNGEONS_PACK = json.load(f)
with open(DATA / "content_rooms.json", encoding="utf-8") as f:
    ROOMS_PACK = json.load(f)
with open(DATA / "content_room_templates.json", encoding="utf-8") as f:
    ROOM_TEMPLATES = json.load(f)

print("Schemas: dungeons.schema.json, rooms.schema.json, room-templates.schema.json")
print("Packs:   content_dungeons.json, content_rooms.json, content_room_templates.json")
print(f"Loaded dungeons keys: {list(DUNGEONS_PACK.keys())}")
print(f"Rooms: {len(ROOMS_PACK.get('rooms', []))}")
print(f"Room templates: {len(ROOM_TEMPLATES.get('templates', ROOM_TEMPLATES.get('roomTemplates', [])))}")
```

## View dungeons and rooms

**Rebalance:** which room has which room feature (room type). **Behaviour:** room feature = room type (training, combat, treasure, rune_forge, etc.).

```python
# Dungeon structure (top-level keys vary by schema)
for key in ["dungeons", "levels", "floors"]:
    if key in DUNGEONS_PACK:
        print(f"DUNGEONS_PACK['{key}']: {len(DUNGEONS_PACK[key])} items")

# Room templates: room type (room feature)
print("\nRoom templates (room type / room feature):")
templates = ROOM_TEMPLATES.get("templates", ROOM_TEMPLATES.get("roomTemplates", []))
for rt in templates:
    print(f"  {rt.get('feature', '')} baseVector keys={list((rt.get('baseVector') or {}).keys())}")

rooms = ROOMS_PACK.get("rooms", [])
print("\nRooms (content_rooms):")
for r in rooms:
    print(f"  {r.get('roomId', '')} feature={r.get('feature', '')}")

dungeons = DUNGEONS_PACK.get("dungeons", [])
if dungeons:
    print("\nFirst dungeon roomBlueprints (first 6):")
    for bp in dungeons[0].get("roomBlueprints", [])[:6]:
        print(f"  {bp.get('roomBlueprintId', '')} {bp.get('name', '')} feature={bp.get('feature', '')}")
```
